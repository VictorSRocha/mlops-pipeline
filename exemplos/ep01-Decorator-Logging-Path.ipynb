### *args ---------------

captura argumentos posicionais:

```python
def soma(a, b):
    return a + b

soma(9,4)
```

```python
def soma(*args):
    num = 0
    for arg in args:
        print(arg)
        num += arg
    return num

soma(6,5,4)
```

### **kwargs ---------------

captura argumentos nomeados:

```python
def soma(**kwargs):
    num = 0
    for chave, valor in kwargs.items():
        print(chave + ': '+ str(valor))
        num += valor
    return num    
    
soma(num1=2, num2=3)
```

### Exemplo Decorator --------------

```python
# Decorator simples para entender o conceito
def ornamento(diz_oi):
    def wrapper(*args, **kwargs):
        print("antes")
        resultado = diz_oi(*args, **kwargs)
        print("depois")
        return resultado
    return wrapper


@ornamento
def diz_oi(nome, sobrenome, teste, ultimo_nome='', ):
    print(f"oi, {nome} - segundo nome {sobrenome}! - ultimo_nome: {ultimo_nome} - variavel: {teste}")


diz_oi("Victor", 'saraiva', 'ok', 'rocha')
```

### pathlib.Path ---------------

```python
from pathlib import Path

# Cria um caminho para a pasta onde salvaremos os dados brutos
RAW_DIR = Path("../data/raw")

# O operador / junta caminhos de forma segura em qualquer sistema operacional
output = RAW_DIR / "customers_api.parquet"

print(output)
# Linux/Mac: data/raw/customers_api.parquet
# Windows:   data\raw\customers_api.parquet

# Cria a pasta se ela ainda não existir
# parents=True cria também as pastas intermediárias, como "data"
# exist_ok=True evita erro caso a pasta já exista
RAW_DIR.mkdir(parents=True, exist_ok=True)

# Verifica se a pasta existe
print("Pasta existe: ", RAW_DIR.exists())
# True

# Verifica se o arquivo existe
print("Arquivo existe: ", output.exists()) # False, se o arquivo ainda não foi criado
print(20*'-')

# Inspeciona partes do caminho
print(output.suffix)  # ".parquet"        -> extensão do arquivo
print(output.stem)    # "customers_api"   -> nome sem extensão
print(output.name)    # "customers_api.parquet" -> nome completo do arquivo
print(output.parent)  # "data/raw"        -> pasta onde o arquivo fica
```
